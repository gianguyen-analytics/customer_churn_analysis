```python
import pandas as pd

df = pd.read_csv("data/Telco-Customer-Churn.csv")

print("Số dòng, số cột:", df.shape)
display(df.head())
df.info()
```

```python
# Đổi TotalCharges sang số; ô không chuyển được sẽ thành NaN
total_charges_numeric = pd.to_numeric(df["TotalCharges"], errors="coerce")

print("Số giá trị TotalCharges không chuyển được sang số:")
print(total_charges_numeric.isna().sum())

print("\nCác dòng cần kiểm tra:")
display(df.loc[total_charges_numeric.isna(),
               ["customerID", "tenure", "MonthlyCharges", "TotalCharges", "Churn"]])
```

```python
# Tạo bản để xử lý, giữ nguyên df đọc từ file gốc
df_clean = df.copy()

# Chuyển TotalCharges từ chữ sang số
df_clean["TotalCharges"] = pd.to_numeric(
    df_clean["TotalCharges"], errors="coerce"
)

# Chỉ điền 0 cho các dòng vừa thiếu TotalCharges vừa có tenure = 0
mask = df_clean["TotalCharges"].isna() & (df_clean["tenure"] == 0)
df_clean.loc[mask, "TotalCharges"] = 0

print("Số dòng được điền 0:", mask.sum())
print("TotalCharges còn thiếu:", df_clean["TotalCharges"].isna().sum())
print("Kiểu dữ liệu mới:", df_clean["TotalCharges"].dtype)
```

```python
print("Số dòng, số cột:", df_clean.shape)
print("customerID bị trùng:", df_clean["customerID"].duplicated().sum())
print("Số dòng trùng hoàn toàn:", df_clean.duplicated().sum())

print("\nSố giá trị thiếu theo từng cột:")
print(df_clean.isna().sum().to_string())
```

```python
print("Các giá trị trong cột Churn:")
print(df_clean["Churn"].value_counts())

total_customers = len(df_clean)
churned_customers = (df_clean["Churn"] == "Yes").sum()
churn_rate = churned_customers / total_customers * 100

print(f"\nTổng khách hàng: {total_customers:,}")
print(f"Khách hàng rời bỏ: {churned_customers:,}")
print(f"Tỷ lệ churn: {churn_rate:.2f}%")
```

```python
churn_by_contract = (
    df_clean.groupby("Contract")
    .agg(
        total_customers=("customerID", "count"),
        churned_customers=("Churn", lambda x: (x == "Yes").sum())
    )
    .reset_index()
)

churn_by_contract["churn_rate"] = (
    churn_by_contract["churned_customers"]
    / churn_by_contract["total_customers"]
    * 100
).round(2)

display(churn_by_contract)
```

Khách dùng hợp đồng theo tháng có tỷ lệ churn 42,71%, cao hơn nhiều so với nhóm hợp đồng một năm (11,27%) và hai năm (2,83%). Trong 1.869 khách rời bỏ, có 1.655 người thuộc nhóm hợp đồng theo tháng. Đây là nhóm nên được ưu tiên tìm hiểu để đề xuất chương trình giữ chân

```python
# Chia khách hàng theo số tháng đã sử dụng dịch vụ
df_clean["tenure_group"] = pd.cut(
    df_clean["tenure"],
    bins=[-1, 12, 24, 48, 72],
    labels=["0–12 tháng", "13–24 tháng", "25–48 tháng", "49–72 tháng"]
)

# Tính số khách và tỷ lệ rời bỏ trong từng nhóm
churn_by_tenure = (
    df_clean.groupby("tenure_group", observed=True)
    .agg(
        total_customers=("customerID", "count"),
        churned_customers=("Churn", lambda values: (values == "Yes").sum())
    )
    .reset_index()
)

churn_by_tenure["churn_rate"] = (
    churn_by_tenure["churned_customers"]
    / churn_by_tenure["total_customers"] * 100
).round(2)

display(churn_by_tenure)
```

Nhóm 0–12 tháng có tỷ lệ churn cao nhất, 47,44%; họ cũng chiếm 1.037 trong 1.869 khách rời bỏ. Đây là nhóm đáng ưu tiên xem xét trong kế hoạch giữ chân khách hàng. Lưu ý đây là so sánh giữa các nhóm trong một bộ dữ liệu, chưa chứng minh rằng cứ sử dụng lâu thêm thì nguy cơ rời bỏ của cùng một khách hàng sẽ giảm.

```python
contract_tenure = (
    df_clean.groupby(["tenure_group", "Contract"], observed=True)
    .agg(
        total_customers=("customerID", "count"),
        churned_customers=("Churn", lambda values: (values == "Yes").sum())
    )
    .reset_index()
)

contract_tenure["churn_rate"] = (
    contract_tenure["churned_customers"]
    / contract_tenure["total_customers"] * 100
).round(2)

display(contract_tenure)
```

Trong từng nhóm thời gian sử dụng, khách dùng hợp đồng theo tháng đều có tỷ lệ churn cao hơn khách dùng hợp đồng một năm hoặc hai năm. Nhóm 0–12 tháng dùng hợp đồng theo tháng ghi nhận 1.024 khách rời bỏ trên 1.994 khách, tương đương 51,35%. Đây là phân khúc nên được ưu tiên tìm hiểu khi đề xuất biện pháp giữ chân. Kết quả cho thấy mối liên hệ trong dữ liệu, chưa chứng minh loại hợp đồng là nguyên nhân gây churn.

```python
churn_by_internet = (
    df_clean.groupby("InternetService")
    .agg(
        total_customers=("customerID", "count"),
        churned_customers=("Churn", lambda values: (values == "Yes").sum())
    )
    .reset_index()
)

churn_by_internet["churn_rate"] = (
    churn_by_internet["churned_customers"]
    / churn_by_internet["total_customers"] * 100
).round(2)

display(churn_by_internet)
```

Khách dùng Fiber optic có tỷ lệ churn 41,89%, cao nhất trong ba nhóm dịch vụ Internet. Nhóm này ghi nhận 1.297 khách rời bỏ. Cần xem thêm loại hợp đồng của họ trước khi đề xuất biện pháp giữ chân.

```python
internet_contract = (
    df_clean.groupby(["Contract", "InternetService"])
    .agg(
        total_customers=("customerID", "count"),
        churned_customers=("Churn", lambda values: (values == "Yes").sum())
    )
    .reset_index()
)

internet_contract["churn_rate"] = (
    internet_contract["churned_customers"]
    / internet_contract["total_customers"] * 100
).round(2)

display(internet_contract)
```

Trong từng loại hợp đồng, khách dùng Fiber optic đều có tỷ lệ churn cao hơn khách dùng DSL. Nhóm Fiber optic dùng hợp đồng theo tháng ghi nhận 1.162/2.128 khách rời bỏ, tương đương 54,61%. Đây là phân khúc cần được ưu tiên tìm hiểu thêm; dữ liệu hiện tại chưa xác định được nguyên nhân rời bỏ.

```python
df_clean["monthly_charge_group"] = pd.cut(
    df_clean["MonthlyCharges"],
    bins=[0, 35, 65, 95, float("inf")],
    labels=["≤35", ">35–65", ">65–95", ">95"]
)

churn_by_charge = (
    df_clean.groupby("monthly_charge_group", observed=True)
    .agg(
        total_customers=("customerID", "count"),
        churned_customers=("Churn", lambda values: (values == "Yes").sum())
    )
    .reset_index()
)

churn_by_charge["churn_rate"] = (
    churn_by_charge["churned_customers"]
    / churn_by_charge["total_customers"] * 100
).round(2)

display(churn_by_charge)
```

Nhóm có MonthlyCharges từ trên 65 đến 95 ghi nhận tỷ lệ churn cao nhất (35,94%) và 936 khách rời bỏ. Tỷ lệ ở nhóm trên 95 là 32,28%, nên dữ liệu không cho thấy churn tăng đều theo mức phí. Cần xét thêm loại dịch vụ Internet trước khi diễn giải chênh lệch này.

```python
internet_charge = (
    df_clean.groupby(
        ["InternetService", "monthly_charge_group"],
        observed=True
    )
    .agg(
        total_customers=("customerID", "count"),
        churned_customers=("Churn", lambda values: (values == "Yes").sum())
    )
    .reset_index()
)

internet_charge["churn_rate"] = (
    internet_charge["churned_customers"]
    / internet_charge["total_customers"] * 100
).round(2)

display(internet_charge)
```

Tỷ lệ churn chung là 26,54%. Các phân khúc nổi bật gồm khách sử dụng 0–12 tháng, khách dùng hợp đồng theo tháng và khách dùng Fiber optic. Nhóm Fiber optic kết hợp hợp đồng theo tháng ghi nhận 1.162 khách rời bỏ trên 2.128 khách (54,61%). Trong khoảng phí >65–95, tỷ lệ churn của Fiber optic (48,81%) cao hơn DSL (7,10%). Đây là các mối liên hệ quan sát được trong dữ liệu; cần tìm hiểu thêm trước khi kết luận nguyên nhân hoặc đưa ra ưu đãi cụ thể.
